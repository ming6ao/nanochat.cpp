# nanochat.cpp launcher (Kaggle)

This is the notebook to import into Kaggle. It is thin. It clones the
repository and runs the real notebook with `%run`.

The real notebook stays `notebooks/nanochat-cpp-on-t4-gpu.ipynb` in the
repository. The repository is the single source of truth. A change to a
content cell takes effect on the next run, so you do not import again.

Import this launcher again only when the launcher itself changes.

Procedure: `docs/host-portability.md`, section 8.

```python
# Clone or refresh the repository. Keep REPO_URL in sync with the real
# notebook at notebooks/nanochat-cpp-on-t4-gpu.ipynb.
import subprocess
from pathlib import Path

REPO_URL = "https://github.com/ming6ao/nanochat.cpp"
REPO = Path("/kaggle/working/nanochat.cpp")

if REPO.is_dir():
    subprocess.run(["git", "-C", str(REPO), "fetch", "--all", "--prune"],
                   check=False)
    subprocess.run(["git", "-C", str(REPO), "pull", "--ff-only"],
                   check=False)
else:
    subprocess.run(["git", "clone", REPO_URL, str(REPO)], check=True)

print("repository:", REPO)
```

```python
# Run the repository notebook in this kernel. The session settings must be
# GPU T4 x2 and Internet on. See docs/host-portability.md section 8.
%run /kaggle/working/nanochat.cpp/notebooks/nanochat-cpp-on-t4-gpu.ipynb
```
